```python
import os
import pandas as pd
import matplotlib.pyplot as plt
cwd= os.getcwd() # current working directory
path = os.path.join(cwd)
```

```python
def get_train_set():
    print ('loading provided train set.')
    fp = os.path.join(path,'train.csv')
    df_train = pd.read_csv(fp, encoding='ISO-8859-1',low_memory=False) 
    return df_train


df_train= get_train_set ()
df_train.head()
```

```python
def get_test_set():    
    print ('loading provided test set .')
    fp = os.path.join(path,'test.csv')
    df_test = pd.read_csv(fp)  # , encoding = 'ISO-8859-1')

    return df_test

df_test= get_test_set()
df_test.head()
# df_test.describe()}
#list(df_test)
```

```python
women=df_train.loc[df_train.Sex=="female"]['Survived']
rate_women=sum(women)/len(women)

print("% of women who survived:", rate_women)
```

```python
men=df_train.loc[df_train.Sex=="male"]['Survived']
rate_men=sum(men)/len(men)

print("% of men who survived:", rate_men)
```

```python
from sklearn.ensemble import RandomForestClassifier
y = df_train['Survived']

features = ['Pclass', "Sex", "SibSp", 'Parch']
X=pd.get_dummies(df_train[features])
X_test =pd.get_dummies(df_test[features])

model=RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
model.fit(X,y)

predications = model.predict(X_test)

output = pd.DataFrame({"passengerId":df_test.PassengerId, "Survived":predications})
output.to_csv("submission.csv",index=False)
```
