```python
import numpy as np

def min_throws(cups, floors):
    dp = np.zeros((floors + 1, cups + 1), dtype=int)

    for k in range(1, floors + 1):
        for e in range(1, cups + 1):
            dp[k][e] = dp[k - 1][e - 1] + dp[k - 1][e] + 1
            if dp[k][e] >= floors:
                return k

print("Par -> 2 cups and 10 floors: ", min_throws(2,10), " good attempts")
print("Par -> 3 cups and 100 floors: ", min_throws(3,100), " good attempts")
print("Par -> 5 cups and 1000 floors: ", min_throws(5,1000), " good attempts")
```
